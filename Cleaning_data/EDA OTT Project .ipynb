## Churn Rate %m

```python
active_subscribers = subscriptions_clean[subscriptions_clean['subscription_status'] == 'Active']['user_id'].nunique()
print("Active subscribers;",active_subscribers)
total_user = subscriptions_clean['user_id'].nunique()
cancelled_user = subscriptions_clean[subscriptions_clean['subscription_status'] == 'Cancelled']['user_id'].nunique()
print("Cancelled User:",cancelled_user)
churn_rate = (cancelled_user / total_user)*100
print("Churn Rate :", round(churn_rate,2) ,"%")

churn_data = pd.Series({
    'Active Users': total_user - cancelled_user,
    'Cancelled Users': cancelled_user
})

churn_data.plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(6,6),
    title='Churn Distribution',
    
)

plt.ylabel("")

plt.show()
```

## Insights
The platform currently maintains a strong active subscriber base with 200,000 active users, indicating healthy overall platform engagement.

A churn rate of 20% is significantly high and suggests that a considerable number of users are discontinuing subscriptions due to possible issues such as low engagement, playback problems, pricing concerns, or weak content retention.

The high number of cancelled users highlights the need for stronger retention strategies, including personalized recommendations, improved streaming quality, and targeted reactivation campaigns to reduce subscriber loss.

# Recommendation CTR Analysis

```python
ctr = (
    recommendation_clean['clicked_flag'].sum()
    /
    recommendation_clean['impression_flag'].sum()
) * 100

print("Recommendation CTR:", round(ctr,2), "%")

ctr_data = pd.Series({
    'Clicked': recommendation_clean['clicked_flag'].sum(),
    'Not Clicked':
    recommendation_clean['impression_flag'].sum()
    -
    recommendation_clean['clicked_flag'].sum()
})

ctr_data.plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(6,6),
    title='Recommendation CTR'
)

plt.ylabel("")

plt.show()
```

# Watch Conversion Analysis

```python
watch_conversion = (
    recommendation_clean['watched_after_click_flag'].sum()
    /
    recommendation_clean['clicked_flag'].sum()
) * 100

print("Watch Conversion:", round(watch_conversion,2), "%")

conversion_data = pd.Series({
    'Watched After Click':
    recommendation_clean[
        'watched_after_click_flag'
    ].sum(),

    'No Watch':
    recommendation_clean[
        'clicked_flag'
    ].sum()
    -
    recommendation_clean[
        'watched_after_click_flag'
    ].sum()
})

conversion_data.plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(6,6),
    title='Watch Conversion Analysis'
)

plt.ylabel("")

plt.show()
```

## Content ROI

```python
watch_movie = watch_history_clean.merge(
    movies_series_clean,
    on='content_id',
    how='left'
)

content_roi = watch_movie.groupby(
    'title'
).agg({
    'production_cost':'mean',
    'watch_duration_minutes':'sum'
})

content_roi['roi_score'] = (
    content_roi['watch_duration_minutes']
    /
    content_roi['production_cost']
)

print(content_roi.head())

top_roi = content_roi.sort_values(
    'roi_score',
    ascending=False
).head(10)

top_roi['roi_score'].plot(
    kind='bar',
    figsize=(12,5),
    title='Top Content ROI',
    color = [
    "blue",
    "green",
    "red",
    "cyan",
    "magenta",
    "yellow",
    "black",
    "orange",
    "purple",
    "brown"]
)

plt.xlabel("Title")

plt.ylabel("ROI Score")

plt.xticks(rotation=45)

plt.show()
```

```python
watch_sample = watch_history_clean.sample(
    500000,
    random_state=42
)
watch_summary = watch_sample.groupby(
    'content_id',
    observed=True
)['watch_duration_minutes'].sum().reset_index()

observed=True

roi_analysis = watch_summary.merge(
    movies_series_clean[
        ['content_id',
         'title',
         'genre',
         'production_cost']
    ],
    on='content_id',
    how='left'
)

roi_analysis['estimated_revenue'] = (
    roi_analysis['watch_duration_minutes']
    * 0.5
)

roi_analysis['roi_score'] = (
    roi_analysis['estimated_revenue']
    /
    roi_analysis['production_cost']
)

top_roi = roi_analysis.sort_values(
    'roi_score',
    ascending=False
).head(10)

print(top_roi)

top_roi.plot(
    x='title',
    y='roi_score',
    kind='bar',
    figsize=(12,5),
    title='Top ROI Content',
    color = [
    "blue",
    "green",
    "red",
    "cyan",
    "magenta",
    "yellow",
    "black",
    "orange",
    "purple",
    "brown"]
)

plt.xlabel("Title")

plt.ylabel("ROI Score")

plt.xticks(rotation=45)

plt.show()
```
