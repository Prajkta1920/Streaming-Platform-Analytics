### gender column

```python
users_clean['gender'] = users_clean['gender'].str.strip().str.title()
users_clean['gender'] = users_clean['gender'].replace('M','Male')
print(users_clean['gender'].unique())
```

```python
male_names = [
    'Amit', 'Ravi', 'Vijay', 'Sanjay', 'Arun',
    'Kiran', 'Suresh', 'Rahul', 'Mahesh',
    'Arjun', 'Naveen', 'Rajesh', 'Vinod',
    'Prasad', 'Raj'
]

female_names = [
    'Priya', 'Sunita', 'Lakshmi', 'Meena',
    'Kavitha', 'Deepa', 'Anita', 'Pooja',
    'Nisha', 'Rekha', 'Divya', 'Swathi',
    'Sneha', 'Usha', 'Suma'
]
```

```python
def assign_gender(name):

    first_name = name.split()[0]

    if first_name in male_names:
        return 'Male'

    elif first_name in female_names:
        return 'Female'

    else:
        return 'Unknown'
```

```python
users_clean['gender'] = users_clean[
    'user_name'
].apply(assign_gender)
```
